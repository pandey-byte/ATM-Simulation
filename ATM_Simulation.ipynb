```python
"""ATM Simulation
Features: PIN login, check balance, deposit, withdraw, mini statement.
"""

balance = 5000.0
PIN = "1234"
transactions = []  # (type, amount, balance_after)


def check_balance():
    print(f"\nCurrent balance: Rs. {balance:.2f}")


def deposit():
    global balance
    try:
        amount = float(input("Enter amount to deposit: Rs. "))
    except ValueError:
        print("Invalid input. Enter a number.")
        return
    if amount <= 0:
        print("Amount must be greater than 0.")
        return
    balance += amount
    transactions.append(("Deposit", amount, balance))
    print(f"Rs. {amount:.2f} deposited successfully.")
    check_balance()


def withdraw():
    global balance
    try:
        amount = float(input("Enter amount to withdraw: Rs. "))
    except ValueError:
        print("Invalid input. Enter a number.")
        return
    if amount <= 0:
        print("Amount must be greater than 0.")
    elif amount > balance:
        print("Insufficient balance.")
    else:
        balance -= amount
        transactions.append(("Withdraw", amount, balance))
        print(f"Please collect your cash: Rs. {amount:.2f}")
        check_balance()


def mini_statement():
    print("\n--- Mini Statement (last 5) ---")
    if not transactions:
        print("No transactions yet.")
    for t, amt, bal in transactions[-5:]:
        print(f"{t:<9} Rs. {amt:>9.2f}   Balance: Rs. {bal:.2f}")


def login():
    for attempt in range(3):
        if input("Enter 4-digit PIN: ") == PIN:
            return True
        print(f"Wrong PIN. {2 - attempt} attempt(s) left.")
    print("Card blocked. Contact your bank.")
    return False


def main():
    print("===== WELCOME TO PYATM =====")
    if not login():
        return
    while True:
        print("\n1. Check Balance\n2. Deposit\n3. Withdraw\n4. Mini Statement\n5. Exit")
        choice = input("Choose an option (1-5): ").strip()
        if choice == "1":
            check_balance()
        elif choice == "2":
            deposit()
        elif choice == "3":
            withdraw()
        elif choice == "4":
            mini_statement()
        elif choice == "5":
            print("Thank you for banking with us. Goodbye!")
            break
        else:
            print("Invalid choice. Try again.")


if __name__ == "__main__":
    main()
```
